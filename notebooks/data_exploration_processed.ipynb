```python
from cropspectai.data_analysis import exploration
from cropspectai import config
from cropspectai.data_analysis.preprocessing import split_dataset_into_X_y
import pandas as pd

from cropspectai.data_analysis import visualization
```

```python
train_path = config.PROCESSED_PARQUET_DATA_SPLITS_PATH / "train.parquet"
test_path = config.PROCESSED_PARQUET_DATA_SPLITS_PATH / "test.parquet"

df_train = pd.read_parquet(train_path)
df_test = pd.read_parquet(test_path)
```

```python
exploration.validate_spectral_values(df_train)
```

```python
exploration.explore_dataset(df_train)
```

```python
exploration.validate_spectral_values(df_test)
```

```python
exploration.explore_dataset(df_test)
```
