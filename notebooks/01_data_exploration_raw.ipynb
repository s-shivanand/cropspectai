### X_train and y_train split (also drop unnamed and dataset columns)

```python
target_columns = ['cab', 'cw', 'cm', 'LAI', 'cp', 'cbc', 'car', 'anth']
y_train = df_train[target_columns]
y_train.describe()
```

```python
X_train = df_train.drop(target_columns + ['Unnamed: 0', 'dataset'], axis=1)
X_train.describe()
```

## check duplicate rows after split

```python
X_train.duplicated().sum()
```

```python
y_train.duplicated().sum()
```

- Identical X (X1 = X2), identical y (y1 = y2) is a true duplicate
- Identical X (X1 = X2), different y (y1 != y2) is not a duplicate training observation

```python
# grouped together so identical rows appear adjacent

X_train[X_train.duplicated(keep=False)].sort_values(by=X_train.columns.tolist())
```

```python
for col in X_train:
    if X_train[col].max() > 1:
        raise ValueError(f"Column '{col}' has values greater than 1.")
    elif X_train[col].min() < 0:
        raise ValueError(f"Column '{col}' has values lesser than 0.")
```
