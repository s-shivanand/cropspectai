```python
from cropspectai.data_analysis import exploration
from cropspectai import ml_config
from cropspectai.data_analysis import preprocessing
import pandas as pd
```

# train data 

```python
train_path = ml_config.DOWNSAMPLED_PARQUET_DATA_SPLITS_PATH / "train.parquet"

df_train = pd.read_parquet(train_path)
```

## validate

```python
exploration.validate_spectral_values(df_train)
```

## exploration

```python
df_train.head()
```

```python
exploration.explore_dataset(df_train)
```

## further info: dulplicates

```python
# see all rows involved in duplication

df_train[df_train.duplicated(keep=False)]
```

```python
# grouped together so identical rows appear adjacent

df_train[df_train.duplicated(keep=False)].sort_values(by=df_train.columns.tolist())
```

## DROP duplicates rows in df_train : full rows + no labels

```python
df_train = preprocessing.drop_duplicates(df_train).copy()
```

```python
df_train.shape
```

## check duplicate rows after split

```python
X_train, y_train = preprocessing.split_dataset_into_X_y(df_train)
```

```python
X_train.duplicated().sum()
```

```python
y_train.duplicated().sum()
```

```python
y_train[y_train.duplicated(keep=False)].sort_values(by=y_train.columns.tolist())
```

```python
print(X_train.loc[496])
print(X_train.loc[582])
```

## save cleaned train data

```python
output_path = ml_config.PROCESSED_PARQUET_DATA_SPLITS_PATH / "all" / "train.parquet"

df_train.to_parquet(
    output_path,
    engine="pyarrow",
    index=False,
)
```
